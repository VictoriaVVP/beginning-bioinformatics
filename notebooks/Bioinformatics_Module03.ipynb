####**TASK P23.1** Modify the script from TASK P22.1 so that it prints the GC content for each sequence in the file

####**TASK P23.2** <font color=orange> **Complete Rosalind #9  Computing GC Content**</font>

````python
## Problem 9 Solution was edited using Gemini

from Bio.SeqIO import parse
from Bio import SeqIO
from Bio.SeqUtils import gc_fraction

max_id = ""
max_gc = 0.0

# 1. Parse each record from your downloaded Rosalind FASTA file
for seq_record in SeqIO.parse("rosalind_gc.txt", "fasta"):
    # Calculate GC fraction (returns a decimal, e.g., 0.609195...)
    current_gc = gc_fraction(seq_record.seq) * 100

    # Keep track of whichever sequence has the highest percentage
    if current_gc > max_gc:
        max_gc = current_gc
        max_id = seq_record.id

# 2. Print output in Rosalind's expected format
print(max_id)
print(f"{max_gc:.6f}")
````
